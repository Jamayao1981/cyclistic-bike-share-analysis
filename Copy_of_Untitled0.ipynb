```python
import pandas as pd

archivos = ['202508-divvy-tripdata.xlsx', '202509-divvy-tripdata.xlsx', '202510-divvy-tripdata.xlsx',
            '202511-divvy-tripdata.xlsx', '202512-divvy-tripdata.xlsx', '202601-divvy-tripdata.xlsx',
            '202602-divvy-tripdata.xlsx', '202603-divvy-tripdata.xlsx', '202604-divvy-tripdata.xlsx',
            '202605-divvy-tripdata.xlsx', '202606-divvy-tripdata.xlsx', '202607-divvy-tripdata.xlsx']

ruta = '/content/drive/MyDrive/Colab Notebooks/'
tablas = []

for archivo in archivos:
    print(f"Leyendo {archivo}...")
    tabla = pd.read_excel(ruta + archivo, engine='calamine')
    print(f"  → {tabla.shape[0]} filas")
    tablas.append(tabla)

datos = pd.concat(tablas, ignore_index=True)
print("TOTAL:", datos.shape)
```

```python
print(datos.columns.tolist())
```

```python
datos = datos.drop(columns=['ride_lenght', 'ride_leght', 'day_of_teh_week',
                              'day_of_the_week', 'day_of_weel', 'day_of_week'])
print(datos.shape)
```

```python
# 1. Corregir fechas (formato mixto)
datos['started_at'] = pd.to_datetime(datos['started_at'], format='mixed', dayfirst=False)
datos['ended_at'] = pd.to_datetime(datos['ended_at'], format='mixed', dayfirst=False)

# 2. Columnas derivadas
datos['mes'] = datos['started_at'].dt.month
datos['duracion_min'] = (datos['ended_at'] - datos['started_at']).dt.total_seconds() / 60
datos['dia_semana'] = datos['started_at'].dt.day_name()

# 3. Filtrar duraciones imposibles (<1 min o >24h)
antes = datos.shape[0]
datos = datos[(datos['duracion_min'] >= 1) & (datos['duracion_min'] <= 1440)]
print(f"Filas eliminadas: {antes - datos.shape[0]}")

# 4. Verificación
print(datos.shape)
print(datos.groupby('mes').size().sort_index())
```

```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
datos = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/cyclistic_consolidado_limpio.csv')
print(datos.shape)
```

```python
resumen_mes = datos.pivot_table(index='mes', columns='member_casual', values='ride_id', aggfunc='count')
resumen_dia = datos['dia_semana'].value_counts()
resumen_duracion_mes = datos.groupby(['mes', 'member_casual'])['duracion_min'].mean().unstack()
top_inicio = datos['start_station_name'].value_counts().head(10)
top_fin = datos['end_station_name'].value_counts().head(10)
resumen_bici = datos['rideable_type'].value_counts()
crosstab_bici = pd.crosstab(datos['rideable_type'], datos['member_casual'], normalize='columns') * 100
duracion_bici = datos.groupby('rideable_type')['duracion_min'].mean()

ruta = '/content/drive/MyDrive/Colab Notebooks/resumenes_cyclistic.xlsx'
with pd.ExcelWriter(ruta) as writer:
    resumen_mes.to_excel(writer, sheet_name='viajes_por_mes')
    resumen_dia.to_excel(writer, sheet_name='viajes_por_dia')
    resumen_duracion_mes.to_excel(writer, sheet_name='duracion_por_mes')
    top_inicio.to_excel(writer, sheet_name='top_estaciones_inicio')
    top_fin.to_excel(writer, sheet_name='top_estaciones_fin')
    resumen_bici.to_excel(writer, sheet_name='tipo_bici')
    crosstab_bici.to_excel(writer, sheet_name='tipo_bici_por_usuario')
    duracion_bici.to_excel(writer, sheet_name='duracion_por_tipo_bici')

print("Guardado en:", ruta)
```
